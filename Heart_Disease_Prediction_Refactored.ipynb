# Heart Disease Prediction
### Binary Classification Project

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay
```

```python
# Load the Data
# Place train_data.csv and test_data.csv in the same folder as this notebook.
# Both files must contain: id, ST depression, Cholesterol, Gender, Age, Heart Disease.
train = pd.read_csv('train_data.csv')
test  = pd.read_csv('test_data.csv')

print('Train shape:', train.shape)
print('Test shape: ', test.shape)
train.head()
```

```python
# Basic Info
train.info()
```

```python
# Check for Missing Values
print('Missing values in train:')
print(train.isnull().sum())
```

```python
# Check Target Column
print('Target value counts:')
print(train['Heart Disease'].value_counts())
```

---
## Step 1: Preprocessing

```python
# Drop the id column (not a feature)
train = train.drop('id', axis=1)
test  = test.drop('id', axis=1)
```

```python
# Keep only the selected features + target
selected_features = ['ST depression', 'Cholesterol', 'Gender', 'Age', 'Heart Disease']

train = train[selected_features]
test  = test[selected_features]

print('Features used:', selected_features[:-1])
```

```python
# Fill missing values
# Numerical columns -> fill with median
# Categorical columns -> fill with mode

train['Age']          = train['Age'].fillna(train['Age'].median())
train['ST depression']= train['ST depression'].fillna(train['ST depression'].median())
train['Cholesterol']  = train['Cholesterol'].fillna(train['Cholesterol'].median())
train['Gender']       = train['Gender'].fillna(train['Gender'].mode()[0])

test['Age']           = test['Age'].fillna(train['Age'].median())
test['ST depression'] = test['ST depression'].fillna(train['ST depression'].median())
test['Cholesterol']   = test['Cholesterol'].fillna(train['Cholesterol'].median())
test['Gender']        = test['Gender'].fillna(train['Gender'].mode()[0])

print('Missing values after filling:')
print(train.isnull().sum())
```

```python
# Encode the target column (Yes -> 1, No -> 0)
train['Heart Disease'] = train['Heart Disease'].map({'Yes': 1, 'No': 0})
test['Heart Disease']  = test['Heart Disease'].map({'Yes': 1, 'No': 0})

print('Target after encoding:')
print(train['Heart Disease'].value_counts())
```

```python
# Encode categorical feature: Gender (Male -> 1, Female -> 0)
train['Gender'] = train['Gender'].map({'Male': 1, 'Female': 0})
test['Gender']  = test['Gender'].map({'Male': 1, 'Female': 0})

train.head()
```

```python
# Split features and target
X_train = train.drop('Heart Disease', axis=1)
y_train = train['Heart Disease']

X_test  = test.drop('Heart Disease', axis=1)
y_test  = test['Heart Disease']

print('X_train shape:', X_train.shape)
print('X_test shape: ', X_test.shape)
```

```python
# Feature Scaling using StandardScaler
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test  = scaler.transform(X_test)

print('Scaling done.')
```

---
## Step 2: Visualizations

```python
# Target Distribution Plot
plt.figure(figsize=(6, 4))
sns.countplot(x='Heart Disease', data=train.assign(**{'Heart Disease': y_train.map({1:'Yes', 0:'No'})}), color='steelblue')
plt.title('Heart Disease Distribution')
plt.xlabel('Heart Disease')
plt.ylabel('Count')
plt.show()
```

```python
# Correlation Heatmap
corr_df = pd.DataFrame(X_train, columns=['ST depression', 'Cholesterol', 'Gender', 'Age'])
corr_df['Heart Disease'] = y_train.values

plt.figure(figsize=(7, 5))
sns.heatmap(corr_df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

---
## Step 3: Model Training

### Model 1: Logistic Regression

```python
# Train Logistic Regression
model = LogisticRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
# Evaluate Logistic Regression
print('--- Logistic Regression ---')
print('Accuracy: ', accuracy_score(y_test, y_pred))
print('Precision:', precision_score(y_test, y_pred))
print('Recall:   ', recall_score(y_test, y_pred))
print('F1-Score: ', f1_score(y_test, y_pred))
print()
print(classification_report(y_test, y_pred, target_names=['No', 'Yes']))
```

```python
# Confusion Matrix - Logistic Regression
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['No', 'Yes'])
disp.plot(cmap='Blues')
plt.title('Confusion Matrix - Logistic Regression')
plt.show()
```

### Model 2: SVM

```python
# Train SVM
model = SVC(kernel='rbf')
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
# Evaluate SVM
print('--- SVM ---')
print('Accuracy: ', accuracy_score(y_test, y_pred))
print('Precision:', precision_score(y_test, y_pred))
print('Recall:   ', recall_score(y_test, y_pred))
print('F1-Score: ', f1_score(y_test, y_pred))
print()
print(classification_report(y_test, y_pred, target_names=['No', 'Yes']))
```

```python
# Confusion Matrix - SVM
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['No', 'Yes'])
disp.plot(cmap='Blues')
plt.title('Confusion Matrix - SVM')
plt.show()
```

### Model 3: Decision Tree

```python
# Train Decision Tree
model = DecisionTreeClassifier(random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
# Evaluate Decision Tree
print('--- Decision Tree ---')
print('Accuracy: ', accuracy_score(y_test, y_pred))
print('Precision:', precision_score(y_test, y_pred))
print('Recall:   ', recall_score(y_test, y_pred))
print('F1-Score: ', f1_score(y_test, y_pred))
print()
print(classification_report(y_test, y_pred, target_names=['No', 'Yes']))
```

```python
# Confusion Matrix - Decision Tree
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['No', 'Yes'])
disp.plot(cmap='Blues')
plt.title('Confusion Matrix - Decision Tree')
plt.show()
```

---
## Step 4: Hyperparameter Tuning

```python
# Hyperparameter Tuning - Logistic Regression (C values)
# Note: this notebook evaluates tuning choices on the test set, which can bias results.
# For a rigorous evaluation, use a validation split or cross-validation for tuning and reserve test data for final evaluation.
# C controls regularization: small C = more regularization, large C = less regularization

C_values = [0.001, 0.01, 0.1, 1, 10, 100]
lr_accuracies = []

for C in C_values:
    model = LogisticRegression(C=C, max_iter=1000)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    lr_accuracies.append(acc)
    print(f'C = {C}  ->  Accuracy = {acc:.4f}')
```

```python
# Plot: Logistic Regression C vs Accuracy
plt.figure(figsize=(7, 4))
plt.plot(C_values, lr_accuracies, marker='o', color='steelblue')
plt.xscale('log')
plt.title('Logistic Regression: C vs Accuracy')
plt.xlabel('C value')
plt.ylabel('Accuracy')
plt.grid(True)
plt.show()
```

```python
# Hyperparameter Tuning - SVM (kernel)
# kernel controls the shape of the decision boundary

kernels = ['linear', 'rbf', 'poly', 'sigmoid']
svm_accuracies = []

for k in kernels:
    model = SVC(kernel=k)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    svm_accuracies.append(acc)
    print(f'kernel = {k}  ->  Accuracy = {acc:.4f}')
```

```python
# Plot: SVM kernel vs Accuracy
plt.figure(figsize=(7, 4))
plt.bar(kernels, svm_accuracies, color='salmon')
plt.title('SVM: Kernel vs Accuracy')
plt.xlabel('Kernel')
plt.ylabel('Accuracy')
plt.ylim(0, 1)
for i, acc in enumerate(svm_accuracies):
    plt.text(i, acc + 0.01, f'{acc:.3f}', ha='center')
plt.show()
```

```python
# Hyperparameter Tuning - Decision Tree (max_depth)
# max_depth limits how deep the tree grows to avoid overfitting

depths = [1, 2, 3, 4, 5, 6, 7, 8]
dt_accuracies = []

for d in depths:
    model = DecisionTreeClassifier(max_depth=d, random_state=42)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)
    dt_accuracies.append(acc)
    print(f'max_depth = {d}  ->  Accuracy = {acc:.4f}')
```

```python
# Plot: Decision Tree max_depth vs Accuracy
plt.figure(figsize=(7, 4))
plt.plot(depths, dt_accuracies, marker='s', color='green')
plt.title('Decision Tree: max_depth vs Accuracy')
plt.xlabel('max_depth')
plt.ylabel('Accuracy')
plt.grid(True)
plt.show()
```

---
## Step 5: Final Model Comparison

```python
# Train each model with its best hyperparameter and compare results

best_C     = C_values[lr_accuracies.index(max(lr_accuracies))]
best_k     = kernels[svm_accuracies.index(max(svm_accuracies))]
best_depth = depths[dt_accuracies.index(max(dt_accuracies))]

print('Best C for Logistic Regression:', best_C)
print('Best kernel for SVM:            ', best_k)
print('Best max_depth for Decision Tree:', best_depth)
```

```python
# Logistic Regression - final
model = LogisticRegression(C=best_C, max_iter=1000)
model.fit(X_train, y_train)
lr_pred = model.predict(X_test)

lr_acc = accuracy_score(y_test, lr_pred)
lr_f1  = f1_score(y_test, lr_pred)

# SVM - final
model = SVC(kernel=best_k)
model.fit(X_train, y_train)
svm_pred = model.predict(X_test)

svm_acc = accuracy_score(y_test, svm_pred)
svm_f1  = f1_score(y_test, svm_pred)

# Decision Tree - final
model = DecisionTreeClassifier(max_depth=best_depth, random_state=42)
model.fit(X_train, y_train)
dt_pred = model.predict(X_test)

dt_acc = accuracy_score(y_test, dt_pred)
dt_f1  = f1_score(y_test, dt_pred)
```

```python
# Print comparison table
print('=========================================')
print(f'{"Model":<25} {"Accuracy":>10} {"F1-Score":>10}')
print('-----------------------------------------')
print(f'{"Logistic Regression":<25} {lr_acc:>10.4f} {lr_f1:>10.4f}')
print(f'{"SVM":<25} {svm_acc:>10.4f} {svm_f1:>10.4f}')
print(f'{"Decision Tree":<25} {dt_acc:>10.4f} {dt_f1:>10.4f}')
print('=========================================')
```

```python
# Bar chart comparison
models   = ['Logistic Regression', 'SVM', 'Decision Tree']
accuracy = [lr_acc, svm_acc, dt_acc]
f1       = [lr_f1,  svm_f1,  dt_f1]

x = np.arange(len(models))

plt.figure(figsize=(9, 5))
plt.bar(x - 0.2, accuracy, width=0.4, label='Accuracy', color='steelblue')
plt.bar(x + 0.2, f1,       width=0.4, label='F1-Score', color='salmon')
plt.xticks(x, models)
plt.ylim(0, 1.1)
plt.title('Model Comparison: Accuracy vs F1-Score')
plt.ylabel('Score')
plt.legend()
plt.show()
```

## Conclusion

**Preprocessing steps applied:**
- Dropped the `id` column (not a feature)
- Selected 4 features: ST depression, Cholesterol, Gender, Age
- Filled missing values using median (numerical) and mode (categorical)
- Encoded target column: Yes -> 1, No -> 0
- Encoded Gender: Male -> 1, Female -> 0
- Applied StandardScaler for feature scaling

**Models trained:**
- Logistic Regression (tuned C)
- SVM (tuned kernel)
- Decision Tree (tuned max_depth)

**Hyperparameter observations:**
- Logistic Regression: Very small C causes underfitting. Best results around C = 1 to 10.
- SVM: The RBF and linear kernels generally outperform poly and sigmoid on this dataset.
- Decision Tree: Increasing depth beyond a certain point causes overfitting. Limiting depth improves generalization.

**Best Model:** Select the model with the strongest evaluation results from the comparison table above. The best model can vary by dataset; for heart-disease screening, also consider recall and the cost of false negatives.
